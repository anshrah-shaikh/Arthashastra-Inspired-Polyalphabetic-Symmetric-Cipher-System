```python
# Arthaśāstra-Inspired Polyalphabetic Symmetric Cipher
# AIPSC

def prepare_key(key):
    """Remove non-alphabetic characters and convert key to uppercase."""
    return ''.join(char.upper() for char in key if char.isalpha())


def encrypt(plaintext, key):
    """Encrypt plaintext using the AIPSC algorithm."""

    key = prepare_key(key)

    if not key:
        raise ValueError("Key must contain at least one alphabetic character.")

    ciphertext = ""
    key_index = 0
    position = 1

    for char in plaintext:

        if char.isalpha():
            # Convert plaintext character to uppercase numerical value
            p = ord(char.upper()) - ord('A')

            # Get corresponding key character cyclically
            k = ord(key[key_index % len(key)]) - ord('A')

            # Positional shift
            s = position

            # Encryption formula:
            # Cᵢ = (Pᵢ + Kᵢ + Sᵢ) mod 26
            c = (p + k + s) % 26

            encrypted_char = chr(c + ord('A'))

            # Preserve original case
            if char.islower():
                encrypted_char = encrypted_char.lower()

            ciphertext += encrypted_char

            key_index += 1
            position += 1

        else:
            # Preserve spaces, numbers and special characters
            ciphertext += char

    return ciphertext


def decrypt(ciphertext, key):
    """Decrypt ciphertext using the AIPSC algorithm."""

    key = prepare_key(key)

    if not key:
        raise ValueError("Key must contain at least one alphabetic character.")

    plaintext = ""
    key_index = 0
    position = 1

    for char in ciphertext:

        if char.isalpha():
            # Convert ciphertext character to numerical value
            c = ord(char.upper()) - ord('A')

            # Get corresponding key character cyclically
            k = ord(key[key_index % len(key)]) - ord('A')

            # Positional shift
            s = position

            # Decryption formula:
            # Pᵢ = (Cᵢ - Kᵢ - Sᵢ) mod 26
            p = (c - k - s) % 26

            decrypted_char = chr(p + ord('A'))

            # Preserve original case
            if char.islower():
                decrypted_char = decrypted_char.lower()

            plaintext += decrypted_char

            key_index += 1
            position += 1

        else:
            # Preserve spaces, numbers and special characters
            plaintext += char

    return plaintext


# --------------------------------------------------
# Main Program
# --------------------------------------------------

print("=" * 63)
print("  ARTHAŚĀSTRA-INSPIRED POLYALPHABETIC SYMMETRIC CIPHER (AIPSC)")
print("=" * 63)

while True:

    print("\n1. Encrypt")
    print("2. Decrypt")
    print("3. Exit")

    choice = input("\nEnter your choice: ")

    if choice == "1":

        plaintext = input("Enter plaintext: ")
        key = input("Enter secret key: ")

        try:
            ciphertext = encrypt(plaintext, key)

            print("\nCiphertext:", ciphertext)

        except ValueError as e:
            print("Error:", e)

    elif choice == "2":

        ciphertext = input("Enter ciphertext: ")
        key = input("Enter secret key: ")

        try:
            plaintext = decrypt(ciphertext, key)

            print("\nDecrypted plaintext:", plaintext)

        except ValueError as e:
            print("Error:", e)

    elif choice == "3":

        print("\nExiting program...")
        break

    else:
        print("\nInvalid choice. Please enter 1, 2 or 3.")
```
